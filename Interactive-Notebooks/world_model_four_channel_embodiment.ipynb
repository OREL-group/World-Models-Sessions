```python
## First attempt at building a world model using four-channel embodiment (Gibsonian Info)

import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ------------------------------------------------------------------
# CHANNEL DEFINITIONS
# ------------------------------------------------------------------

channels = {
    "Vision (V)": {
        "vector": np.array([0.0, 1.0]),
        "strength": 1.0,
        "cue": "Forward Motion",
        "color": "#1f77b4"
    },
    "Haptic (H)": {
        "vector": np.array([0.15, 0.9]),
        "strength": 0.9,
        "cue": "Forward Momentum",
        "color": "#ff7f0e"
    },
    "Proprioception (P)": {
        "vector": np.array([-0.8, 0.0]),
        "strength": 0.7,
        "cue": "Shift Left",
        "color": "#2ca02c"
    },
    "Auditory (A)": {
        "vector": np.array([0.0, -0.6]),
        "strength": 0.6,
        "cue": "Birds Chirping",
        "color": "#d62728"
    }
}

# ------------------------------------------------------------------
# GIBSONIAN WORLD MODEL
# ------------------------------------------------------------------

vectors = np.array(
    [v["vector"] * v["strength"] for v in channels.values()]
)

# Direct perception:
# Sum of all active sensory gradients
direct_perception = vectors.sum(axis=0)

# Coherent movement
# Degree to which all channels point together
norms = np.linalg.norm(vectors, axis=1)
unit_vectors = vectors / norms[:, None]

pairwise = []
for i in range(len(unit_vectors)):
    for j in range(i + 1, len(unit_vectors)):
        pairwise.append(np.dot(unit_vectors[i], unit_vectors[j]))

coherence = np.mean(pairwise)

# Disjoint distribution
# Incongruence between channels
disjointness = 1.0 - ((coherence + 1) / 2)

# Inferred perception:
# weighted consensus generated from multisensory integration
inferred = direct_perception * (1 - disjointness / 2)

# ------------------------------------------------------------------
# INTERACTIVE VISUALIZATION
# ------------------------------------------------------------------

fig = make_subplots(
    rows=1,
    cols=2,
    specs=[[{"type": "xy"}, {"type": "indicator"}]],
    subplot_titles=(
        "Four-Channel Embodied World Model",
        "GI Metrics"
    )
)

# Observer at origin
fig.add_trace(
    go.Scatter(
        x=[0],
        y=[0],
        mode="markers+text",
        marker=dict(size=18, color="black"),
        text=["Observer"],
        textposition="bottom center",
        showlegend=False
    ),
    row=1,
    col=1
)

# Sensory channels
for name, data in channels.items():

    vec = data["vector"] * data["strength"]

    fig.add_trace(
        go.Scatter(
            x=[0, vec[0]],
            y=[0, vec[1]],
            mode="lines+markers+text",
            line=dict(width=5, color=data["color"]),
            marker=dict(size=8),
            text=["", name],
            textposition="top center",
            name=name,
            hovertemplate=(
                f"<b>{name}</b><br>"
                f"Cue: {data['cue']}<br>"
                f"Strength: {data['strength']}"
                "<extra></extra>"
            )
        ),
        row=1,
        col=1
    )

# Direct perception vector
fig.add_trace(
    go.Scatter(
        x=[0, direct_perception[0]],
        y=[0, direct_perception[1]],
        mode="lines+text",
        line=dict(color="black", width=8),
        text=["", "Direct Perception"],
        textposition="middle right",
        name="Direct Perception"
    ),
    row=1,
    col=1
)

# Inferred perception vector
fig.add_trace(
    go.Scatter(
        x=[0, inferred[0]],
        y=[0, inferred[1]],
        mode="lines+text",
        line=dict(color="purple", width=5, dash="dash"),
        text=["", "Inferred Perception"],
        textposition="bottom right",
        name="Inferred Perception"
    ),
    row=1,
    col=1
)

# Gauge: coherence
fig.add_trace(
    go.Indicator(
        mode="gauge+number",
        value=float((coherence + 1) / 2),
        title={"text": "Coherent Movement"},
        gauge={
            "axis": {"range": [0, 1]},
            "bar": {"color": "darkblue"},
            "steps": [
                {"range": [0, 0.4], "color": "lightcoral"},
                {"range": [0.4, 0.7], "color": "khaki"},
                {"range": [0.7, 1.0], "color": "lightgreen"},
            ]
        }
    ),
    row=1,
    col=2
)

fig.update_xaxes(range=[-1.5, 1.5], row=1, col=1)
fig.update_yaxes(range=[-1.2, 1.5], row=1, col=1)

fig.update_layout(
    height=700,
    width=1200,
    title=(
        "World Model of Four-Channel Embodiment "
        "(Gibsonian Information)"
    ),
    legend=dict(
        orientation="h",
        x=0.1,
        y=-0.1
    )
)

# ------------------------------------------------------------------
# REPORT
# ------------------------------------------------------------------

print("\nWORLD MODEL SUMMARY")
print("-------------------")
print(f"Coherent Movement: {coherence:.3f}")
print(f"Disjoint Distribution: {disjointness:.3f}")
print(f"Direct Perception: {direct_perception}")
print(f"Inferred Perception: {inferred}")

fig.show()
```

```python
import numpy as np
import plotly.graph_objects as go

# =====================================================
# FOUR-CHANNEL ACTIVE INFERENCE EMBODIMENT
# =====================================================

np.random.seed(42)

N = 10000
N_STATES = 8

# -----------------------------------------------------
# empirical distributions from pseudo-data
# -----------------------------------------------------

vision_strengths = np.array([
0.01,0.03,0.18,0.11,0.23,0.26,0.28,0.33,0.21,0.19,
0.30,0.26,0.35,0.38,0.40,0.44,0.47,0.49,0.52,0.55,
0.56,0.58,0.61,0.66,0.68
])

angles = np.array([
20,10,-30,-55,-70,-90,-115,-80,-40,-65,
-175,-110,-45,-5,35,75,110,130,150,
175,-175,180,155,130,170,-150,-130,-120,-100
])

# -----------------------------------------------------
# MDP definition
# -----------------------------------------------------

A = np.random.rand(N_STATES, N_STATES)
A = A / A.sum(axis=0)

B = np.random.rand(N_STATES, N_STATES)
B = B / B.sum(axis=0)

preferences = np.ones(N_STATES)

# -----------------------------------------------------
# active inference update
# -----------------------------------------------------

def softmax(x):

    x = x - np.max(x)

    ex = np.exp(x)

    return ex / ex.sum()


def infer_state(observation):

    likelihood = A[observation]

    q = softmax(np.log(likelihood + 1e-8))

    return q


def free_energy(q, likelihood):

    return np.sum(
        q * (
            np.log(q + 1e-8)
            - np.log(likelihood + 1e-8)
        )
    )


# -----------------------------------------------------
# Monte Carlo simulation
# -----------------------------------------------------

channel_names = [
    "Vision",
    "Audition",
    "Haptic",
    "Proprioception"
]

channel_colors = [
    "royalblue",
    "red",
    "orange",
    "green"
]

integrated_angles = []
integrated_strengths = []
integrated_surprisal = []

for trial in range(N):

    vectors = []
    posteriors = []

    for c in range(4):

        signal = np.random.choice(vision_strengths)

        angle = np.random.choice(angles)

        obs = np.random.randint(N_STATES)

        q = infer_state(obs)

        posteriors.append(q)

        x = signal * np.cos(np.radians(angle))
        y = signal * np.sin(np.radians(angle))

        vectors.append([x,y])

    posteriors = np.array(posteriors)

    # -----------------------------------------
    # FREE ENERGY MINIMIZATION BETWEEN CHANNELS
    # -----------------------------------------

    consensus = posteriors.mean(axis=0)

    consensus = consensus / consensus.sum()

    cross_surprisal = 0

    for q in posteriors:

        cross_surprisal += np.sum(
            q * (
                np.log(q+1e-8)
                - np.log(consensus+1e-8)
            )
        )

    vectors = np.array(vectors)

    integrated = vectors.mean(axis=0)

    angle = np.degrees(
        np.arctan2(
            integrated[1],
            integrated[0]
        )
    )

    magnitude = np.linalg.norm(integrated)

    integrated_angles.append(angle)

    integrated_strengths.append(magnitude)

    integrated_surprisal.append(cross_surprisal)

# -----------------------------------------------------
# POLAR FIELD
# -----------------------------------------------------

fig = go.Figure()

fig.add_trace(

    go.Scatterpolar(

        r=np.array(integrated_strengths),

        theta=np.array(integrated_angles),

        mode="markers",

        marker=dict(

            size=4,

            color=np.array(integrated_surprisal),

            colorscale="Viridis",

            colorbar=dict(

                title="Free Energy"

            ),

            opacity=0.45

        ),

        name="Integrated Perception Field"

    )

)

fig.update_layout(

    polar=dict(

        radialaxis=dict(range=[0,1])

    ),

    title=(
        "Four-Channel Embodiment\n"
        "Active Inference + MDP + Free Energy Minimization\n"
        "10,000 Simulated Trials"
    ),

    height=900,

    width=900

)

fig.show()

# -----------------------------------------------------
# SUMMARY
# -----------------------------------------------------

print(
    "Mean Free Energy:",
    np.mean(integrated_surprisal)
)

print(
    "Mean Integrated Signal:",
    np.mean(integrated_strengths)
)
```
